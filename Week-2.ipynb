```python
import pandas as pd

# 1. Load the CSV
df = pd.read_csv("50_startups.csv")

# Print the shape
print("Shape:", df.shape)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])


# 2. Display information about the data
print("\nData Information:")
df.info()

print("\nData Types:")
for column in df.columns:
    print(column, ":", df[column].dtype)


# 3. Check for missing values
print("\nMissing Values:")
print(df.isnull().sum())

missing = df.isnull().sum()

for column in df.columns:
    if missing[column] > 0:
        print(column, "has", missing[column], "missing values")


# 4. Series and DataFrame
series = df["Profit"]
dataframe = df[["Profit"]]

print("\nSeries:")
print(type(series))
print("Shape:", series.shape)

print("\nDataFrame:")
print(type(dataframe))
print("Shape:", dataframe.shape)


# 5. Describe numerical columns
print("\nDescriptive Statistics:")
print(df.describe())

print("\nMean Values:")
print(df.mean(numeric_only=True))

print("\nMaximum Values:")
print(df.max(numeric_only=True))


# 6. Export DataFrame to JSON
df.to_json("startups.json", orient="records")

print("\nJSON file created successfully.")

# Open and display the JSON file
with open("startups.json", "r") as file:
    print(file.read())


# 7. Data points that would not help predict a student's grade
print("\nExamples of data that would not help predict a student's grade:")
print("1. Favorite color")
print("2. Shoe size")
print("3. Phone model")
```
