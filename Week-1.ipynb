```python
ages = [19, 22, 21, 35, 28, 24, 31, 19, 27, 40]

import statistics

# a. Number of people
number_of_people = len(ages)

# b. Total of all ages
total_age = sum(ages)

# c. Mean age
mean_age = total_age / number_of_people

# d. Youngest and oldest
youngest = min(ages)
oldest = max(ages)

# e. Median
median_age = statistics.median(ages)

# Print the answers
print("Number of people:", number_of_people)
print("Total of all ages:", total_age)
print("Mean age:", round(mean_age, 1))
print("Youngest age:", youngest)
print("Oldest age:", oldest)
print("Median age:", median_age)

if mean_age != median_age:
    print("The mean and median are different.")
    print("This suggests that some ages are pulling the average up or down.")
else:
    print("The mean and median are the same.")
```
